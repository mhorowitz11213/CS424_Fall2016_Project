```python
#To make it easier to work with data lets get rid of whitespaces in variable names and extra spaces that may appear
horowitz_data.columns = horowitz_data.columns.str.strip()
sharma_data.columns = sharma_data.columns.str.strip()
rai_data.columns = rai_data.columns.str.strip()
```

```python
#Print column names
print("Horowitz Data")
print(horowitz_data.columns)
print("Sharma Data")
print(sharma_data.columns)
print("Rai Data")
print(rai_data.columns)
```

```python
#Lets get all the data into same data types
#Horowitz--lets get observation previous and actual next into int64
#First we need to convert NA for these two variables into 999--this is a jibberish answer

horowitz_data['Observation Previous (in mins)']=horowitz_data['Observation Previous (in mins)'].fillna(999)
horowitz_data['Observation Next (in Mins), Based on Ventra App']=horowitz_data['Observation Next (in Mins), Based on Ventra App'].fillna(999)
horowitz_data['Actual Next (in Mins)']=horowitz_data['Actual Next (in Mins)'].fillna(999)



#Now lets convert to int64
horowitz_data['Observation Previous (in mins)']=horowitz_data['Observation Previous (in mins)'].astype('int64')
horowitz_data['Observation Next (in Mins), Based on Ventra App']=horowitz_data['Observation Next (in Mins), Based on Ventra App'].astype('int64')
horowitz_data['Actual Next (in Mins)']=horowitz_data['Actual Next (in Mins)'].astype('int64')

#Sharma Data
#First lets get rid of any potential NA with gibberish 999
sharma_data['Observation Previous (in mins)']=sharma_data['Observation Previous (in mins)'].fillna(999)
sharma_data['Observation Next (in Mins), Based on Ventra App']=sharma_data['Observation Next (in Mins), Based on Ventra App'].fillna(999)
sharma_data['Actual Next (in Mins)']=sharma_data['Actual Next (in Mins)'].fillna(999)

#Now lets convert to int64
sharma_data['Observation Previous (in mins)']=sharma_data['Observation Previous (in mins)'].astype('int64')
sharma_data['Observation Next (in Mins), Based on Ventra App']=sharma_data['Observation Next (in Mins), Based on Ventra App'].astype('int64')
sharma_data['Actual Next (in Mins)']=sharma_data['Actual Next (in Mins)'].astype('int64')

#Rai data--
rai_data['Observation Previous (in mins)']=rai_data['Observation Previous (in mins)'].fillna(999)
rai_data['Actual Next (in Mins)']=rai_data['Actual Next (in Mins)'].fillna(999)

#Because the last attribute is object we will also coerce--https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.to_numeric.html
rai_data['Actual Next (in Mins)'] = (
    pd.to_numeric(rai_data['Actual Next (in Mins)'], errors='coerce')
    .fillna(999)
    .astype('int64')
)

rai_data['Observation Previous (in mins)']=rai_data['Observation Previous (in mins)'].astype('int64')
rai_data['Actual Next (in Mins)']=rai_data['Actual Next (in Mins)'].astype('int64')


```

Now lets merge the datasets together

```python
#First we need to make sure the column names are all the same
#1. Observation ID--make this ID
horowitz_data.rename(columns={'Observation ID':'ID'}, inplace=True)
#2. Date is fine
#3. Day of week is fine
#4. Route is fine
#5. Horowitz Stop--Stop-Street
horowitz_data.rename(columns={'Stop':"Stop-Street"}, inplace=True)
#6. Stop-ID needs to become Stop-CTA ID
horowitz_data.rename(columns={'Stop ID': 'Stop-CTA ID'}, inplace=True)
#7. Direction fine
#8. Ventra Time fine
#9. Real Time Fine
#10. Observation Previous Fine
#11. Observation Next fine
#12. Actual next fine
#13. Comments fine
```

We are ready to merge, after we merge we will check for missing values or any other anomalies


```python
#Merge Horowitz and Sharma so that Sharma records are beneath Horowitz
temp_df1=pd.concat([horowitz_data, sharma_data])
temp_df2=pd.concat([temp_df1, rai_data])
print(temp_df2)
#Shape is 102x13--as expected
```

```python
#We now need to check and make sure the dataframe is okay. First we will move the extraneous variables
print(temp_df2.columns)
#Lets inspect each column
#1. ID
print(set(temp_df2['ID']))
#2. Date
print(set(temp_df2['Date']))
#3. Day of Week
print(set(temp_df2['Day of Week']))
#4. Route
print(set(temp_df2['Route']))
#5. Stop-Street
print(set(temp_df2['Stop-Street']))

temp_df2['Stop-Street']=temp_df2['Stop-Street'].str.replace('&', 'and')
print(set(temp_df2['Stop-Street']))
#6. Stop-CTA ID--note the 0 still exists as a gibberish stop number because CTA does not print on stop stop numbers for Loop Link Stops. Will correct later.
print(set(temp_df2['Stop-CTA ID']))
#7. Direction
print(set(temp_df2['Direction']))
#8. Ventra time
print(set(temp_df2['Ventra Time (Calculated when previous bus arrives)']))
#9. Real Time
print(set(temp_df2['Real Time']))
#10. Observation Previous
print(set(temp_df2['Observation Previous (in mins)']))
#11. Observation Next
print(set(temp_df2['Observation Next (in Mins), Based on Ventra App']))
#12. Actual Next
print(set(temp_df2['Actual Next (in Mins)']))
#13. Comments
print(set(temp_df2['Comments']))
```

Lets clean up the Stop 0--we will go to the Stop File provided by CTA and check the stop codes and geolocations per stop

```python
#Get incorrect stops
incorrect_stops=temp_df2[temp_df2['Stop-CTA ID']==0]
print(incorrect_stops)
```

```python
#Looking above the incorrect coded ones should only be for loop link--that is true
#Enter correct codes
temp_df2.loc[temp_df2['Stop-Street'] == 'Clark and Washington', 'Stop-CTA ID'] = 18122
temp_df2.loc[temp_df2['Stop-Street'] == 'La Salle and Madison', 'Stop-CTA ID'] = 18123
```

```python
#Check to see this update worked
print(temp_df2)
print(set(temp_df2['Stop-CTA ID']))
```

Now lets append the geo coding

```python
#Read in txt file--see secondary sources zipped folder for the stops.txt file
stops_file=pd.read_csv('stops.txt',sep=',', header=0)
print(stops_file)
```

```python
#Grab just the stop id and the latitutde and logitude from stops_file
stops_file_merger=stops_file[['stop_id','stop_lat','stop_lon']]
print(stops_file_merger)
```

```python
#Rename the stop_id in the merged file to Stop-CTA ID
stops_file_merger.rename(columns={'stop_id':'Stop-CTA ID'}, inplace=True)
print(stops_file_merger)
```

```python
temp_df3=pd.merge(temp_df2,stops_file_merger, on='Stop-CTA ID', how='inner', suffixes=('_original', '_CTAFile'))
print(temp_df3)
```

```python
#Lets check--https://www.geeksforgeeks.org/python/zip-in-python/
print(set(zip(temp_df3['Stop-Street'],temp_df3['stop_lat'],temp_df3['stop_lon'])))

#Visually checked by plotting into Google Maps, all good
assignment2_df=temp_df3
```
